### Descriptive Statistics
Descriptive statistics are statistical methods used to summarize, organize, and describe the main features of a dataset. They provide a simple way to understand large amounts of data by presenting it in a meaningful form using numbers, tables, and graphs.

### Measures of Central Tendency

Measures of central tendency indicate the central or typical value of a dataset. They describe where most of the data values are concentrated.

1. Mean (Arithmetic Mean)

The mean is the average value of a dataset and is calculated by dividing the sum of all observations by the total number of observations.

2. Median

The median is the middle value of a dataset when the values are arranged in ascending or descending order.

If the number of observations is odd, the median is the middle value.
If even, it is the average of the two middle values.
    
3. Mode

The mode is the most frequently occurring value in a dataset. A dataset may have one mode, more than one mode, or no mode.

### Measures of Variability (Dispersion)

Measures of variability describe the spread or dispersion of data values around the central value.

1. Range

Range is the difference between the maximum and minimum values in a dataset.

Range=Maximum−Minimum

2. Variance

Variance measures how far each data point is from the mean. It is calculated as the average of squared differences from the mean.

3. Standard Deviation

Standard deviation is the square root of variance and indicates the average deviation of data points from the mean.

# Descriptive Statistics using Mall Customers Dataset

## Import required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the dataset

```python
df= pd.read_csv('Mall_Customers.csv')
```

## Display dataset

```python
df.head()
```

## Data Preprocessing

```python
# Display column names
df.columns
```

```python
# Display data types
df.dtypes
```

```python
# Check for null values
df.isnull().sum()
```

## DESCRIPTIVE STATISTICS

```python
# Define bins (age ranges)
bins = [0,20,30,40,50,60,100]
# Define labels for each bin
labels = [ "Teen", "Young Adult","Adult","Middle Age","Senior","Elder"]
df['Age_group']= pd.cut(df['Age'],bins=bins,labels=labels, right=True)
df[['Age','Age_group']].head()
```

```python
# Calculate standard deviation of Spending Score
std_dev = df.groupby('Age_group', observed=False)['Spending Score (1-100)'].std()

print("\nStandard Deviation by Age Group:")
print(std_dev)
```

```python
# Group by Age_Group
grouped = df.groupby('Age_group', observed=False)['Spending Score (1-100)']

# Calculate summary statistics
summary = grouped.agg(['mean', 'median', 'min', 'max', 'std'])

print("\nSummary Statistics:")
print(summary)
```

```python
# Detailed statistics
print("\nFull Description:")
print(grouped.describe())
```

## Group Data by Categorical Variable

```python
# create the group
grouped = df.groupby('Gender')['Spending Score (1-100)']
```

```python
# summary of group
summary_stats = grouped.agg(['mean','median','min','max','std'])
print("Group Statisctics :\n",summary_stats)
```

```python
# Detailed statistics
grouped.describe()
```
