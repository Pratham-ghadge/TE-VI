```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Mall_Customers.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['Gender'].unique()
```

```python
grouped= df.groupby('Gender')
```

```python
stats_summary = grouped.agg(['mean','max','min','std'])
print(stats_summary)
```

```python
grouped.describe().T
```

```python
bins = [0,20,30,40,50,60,100]
lables =['Teen','Young Adult','Adult','Middle Age','Senior','old']
df['Age_group'] = pd.cut( df['Age'],bins= bins,labels=lables)
```

```python
df.head()
```

```python
grouped = df.groupby('Age_group')[['Age','Annual Income (k$)','Spending Score (1-100)'	]]
grouped.agg(['mean','std','count','min','max']).T
```
