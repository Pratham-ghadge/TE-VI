```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk.corpus import wordnet
```

## Download NLTK Data

```python
# Download required resources (run once)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averged_preceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
```

## Sample Document

```python
# Sample text document
text = "Text analytics is the process of analyzing text data to extract meaningful information."
print("Original Text:\n", text)
```

## Tokenization:

```python
# splitting text into words
tokens = word_tokenize(text)
print("\nTokens:\n", tokens)
```

## POS Tagging

```python
# POS Tagging: identify grammatical role of each word
pos_tag = pos_tag(tokens)
print(pos_tag)
```

## Stopwords Removal

```python
stop_words = set(stopwords.words('english'))
stop_word = [word for word in tokens if word.lower() not in stop_words]
print(stop_word)
```

## Stemming

```python
stemmer = PorterStemmer()
for word in stop_word :
    print(word ,"--->",stemmer.stem(word))
```

```python
stem_word = [ stemmer.stem(word) for word in stop_word]
print(stem_word)
```

## Lemmatization

```python
lemmatizer = WordNetLemmatizer()
for word in stop_word:
    print(word,'--->',lemmatizer.lemmatize(word))
```

```python
def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

print("\nLemmatization:")
lemma =[]
for word, tag in pos_tag:
    lemma.append(lemmatizer.lemmatize(word, get_wordnet_pos(tag)))
print(lemma)

lemmatizer.lemmatize('going',pos='v')
```

## Term Frequency (TF)

```python
tf_dict = {}
total_words = len(lemma)
for word in lemma:
    if word in tf_dict:
        tf_dict[word] =+ 1
    else:
      tf_dict[word]=1

print('Term Frequency:\n')
for word in tf_dict:
    tf = tf_dict[word]/ total_words
    print(word,'-->',tf)
    
```

## TF-IDF

```python
# Import libraries
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

# Sample documents
docs = [
    "Text analytics is the process of analyzing text data",
    "Text data is useful for extracting information",
    "Analytics helps in understanding data"
]

# ------------------ TF ------------------
# Count words
cv = CountVectorizer()
tf = cv.fit_transform(docs)

print("TF (Word Count):\n")
print(pd.DataFrame(tf.toarray(), columns=cv.get_feature_names_out()))


# ------------------ TF-IDF ------------------
# Directly calculate TF-IDF
tfv = TfidfVectorizer()
tfidf = tfv.fit_transform(docs)

print("\nTF-IDF:\n")
print(pd.DataFrame(tfidf.toarray(), columns=tfv.get_feature_names_out()))


# ------------------ IDF ------------------
# Show IDF values
print("\nIDF Values:\n")
for word, value in zip(tfv.get_feature_names_out(), tfv.idf_):
    print(word, ":", round(value, 2))
```
