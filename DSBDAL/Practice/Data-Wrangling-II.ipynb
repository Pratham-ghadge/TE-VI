## Handle Missing Values

```python
# check the missing value percentage
df.isnull().sum()/df.shape[0]
```

```python
# In the Dataste column contain more than 50% NULL values the remove column using this df = df.drop(columns='col_name')
df = df.drop(columns='Self_Employed')
```

```python
df.shape
```

```python
# Remove rows  that contains NULL values
df = df.dropna()
```

```python
df.shape
```

```python
#Filling Missing value with the one globle value
df.fillna(10)
```

```python
# Filling Missing Values using backwaed & forward filling
# use axis=1 for left to right filling
df.bfill(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.ffill(inplace=True)
```

```python
df.isnull().sum()
```

```python
# Filling missing value Using Mean for the numerical columns and Mode For Categorical columns
# filling categorical columns
for i in df.select_dtypes(include='str').columns:
    df[i]= df[i].fillna(df[i].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
#filling the numerical columns
for i in df.select_dtypes(include=['float64','int64']).columns:
    df[i]= df[i].fillna(df[i].mean(),inplace=True)
```

```python
df.isnull().sum()
```

```python
# replace() Used to change wrong / inconsistent values into correct ones
df['Gender'] = df['Gender'].replace({'Male':'M','Female':'F'})
```

```python
#replace Same as fillna(), but less preferred
df['Gender'] = df['Gender'].replace(np.nan, df['Gender'].mode()[0])
```

## Outlier Detection

```python
# Boxplot to detect outliers
sns.boxplot(df['ApplicantIncome'])
```

```python
# select all numerical columns for Outlier detection
plt.figure(figsize=(16,20))
col = df.select_dtypes(include=['float64','int64']).columns
for i ,column in enumerate(col,1):
    plt.subplot(3,3,i)
    sns.boxplot(df[column])
    plt.title(column)
plt.tight_layout()
plt.show()
```

```python
# Detect Outliers Using the IQR
df['ApplicantIncome'] = df['ApplicantIncome'].astype('float')
Q1 = df['ApplicantIncome'].quantile(0.25)
Q3 = df['ApplicantIncome'].quantile(0.75)

IQR = Q3 - Q1

l_range = Q1 - 1.5 * IQR
h_range = Q3 + 1.5 * IQR

# Find outliers
outliers = df[ (df['ApplicantIncome'] < l_range )| (df['ApplicantIncome'] > h_range)]

print("Outliers:\n",outliers['ApplicantIncome'])             
```

```python
# remove Outliers using IQR
df =df[(df['ApplicantIncome'] >= l_range) & (df['ApplicantIncome']<= h_range)]
```

```python
# verify using Boxplot
sns.boxplot(df['ApplicantIncome'])
```

```python
# Detect Outliers using Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(df.index,df['ApplicantIncome'])
plt.show()
```

```python
# Detect Outliers using Histogram (Normal Distribution Check)
plt.figure(figsize=(6,4))
sns.histplot(df['ApplicantIncome'],kde=True)
plt.show()
```

```python
sns.boxplot(df['CoapplicantIncome'])
```

```python
# Z-Score Method (Detection)
# Z-score = (X - mean) / standard deviation
from scipy import stats
import numpy as np

z_scores = np.abs(stats.zscore(df['CoapplicantIncome']))

# Show outliers
outliers = df[z_scores > 3]
print(outliers['CoapplicantIncome'])
```

```python
df = df[z_scores < 3]
```

## Data Transformation

```python
# Check skewness before transformation
for i in df.select_dtypes(include=['int64','float64']).columns:
    print(i ,":",df[i].skew())
```

```python
# Apply log transformation
for i in df.select_dtypes(include=['int64','float64']).columns:
    df[i]=np.log(df[i])
for i in df.select_dtypes(include=['int64','float64']).columns:
    print(i ,":",df[i].skew())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Select numeric columns
num_cols = df.select_dtypes(include=['int64','float64']).columns

for col in num_cols:
    
    print(f"\nColumn: {col}")
    
    # Handle zero/negative values safely
    if (df[col] <= 0).any():
        data = df[col] + 1   # shift to avoid log(0)
    else:
        data = df[col]
    
    # Transformations
    log_data = np.log(data)
    log10_data = np.log10(data)
    sqrt_data = np.sqrt(data)
    
    # Print skewness
    print("Original Skew:", df[col].skew())
    print("Log Skew:", log_data.skew())
    print("Log10 Skew:", log10_data.skew())
    print("Sqrt Skew:", sqrt_data.skew())
    
    # Plot histograms
    plt.figure(figsize=(12,8))
    
    plt.subplot(2,2,1)
    sns.histplot(df[col], kde=True)
    plt.title(f"{col} - Original")
    
    plt.subplot(2,2,2)
    sns.histplot(log_data, kde=True)
    plt.title(f"{col} - Log")
    
    plt.subplot(2,2,3)
    sns.histplot(log10_data, kde=True)
    plt.title(f"{col} - Log10")
    
    plt.subplot(2,2,4)
    sns.histplot(sqrt_data, kde=True)
    plt.title(f"{col} - Sqrt")
    
    plt.tight_layout()
    plt.show()
```
