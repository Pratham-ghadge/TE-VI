```python
import pandas as pd
import numpy as np
```

```python
file = open('sample.txt','r')
content = file.read()
print(content)
```

```python
import nltk 
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python

sentence_tokens = sent_tokenize(content)
for i in sentence_tokens:
    print(i)
```

```python
word_tokens = word_tokenize(content)
print(word_tokens)
```

```python
len(word_tokens)
```

```python
## post tagging
```

```python
from nltk import pos_tag
pos_tags = pos_tag(word_tokens)
print(pos_tags)
```

```python
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
```

```python
print(stop_words)
```

```python
word_tokens =[word for word in word_tokens if word.lower() not in stop_words]
print(word_tokens)
```

```python
len(word_tokens)
```

```python
from nltk.stem import PorterStemmer,WordNetLemmatizer

stemmer = PorterStemmer()

sem_word =[]
print('word ---> stemming word')
for word in word_tokens:
    print(word ,'---->',stemmer.stem(word))
    sem_word.append(stemmer.stem(word))

print(sem_word)
```

```python
from nltk.stem import PorterStemmer,WordNetLemmatizer

stemmer =WordNetLemmatizer()

lem_word =[]
print('word ---> lemmatizing word')
for word in word_tokens:
    print(word ,'---->',stemmer.lemmatize(word))
    lem_word.append(stemmer.lemmatize(word))

print(lem_word)
```

```python
print(pos_tags)
```

```python
from nltk.corpus import wordnet
def get_pos_tag(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

    
```

```python
lemma=[]
for word,tag in pos_tags:
    print(word,'--->',stemmer.lemmatize(word,pos= get_pos_tag(tag)))
    lemma.append(stemmer.lemmatize(word,pos= get_pos_tag(tag)))
```

```python
print(lemma)
```

```python
docs = [ 
    'Natural Language Processing (NLP) is a field of Artificial Intelligence',
    'Natural Language Processing is used for voice Assistant',
    'Artificial Intelligence is the growing faster faster'
]
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer

vectorizer =CountVectorizer()
tf = vectorizer.fit_transform(docs)
term_freq = pd.DataFrame(tf.toarray(),columns= vectorizer.get_feature_names_out()) 
print(term_freq)
```

```python
vectorizer =TfidfVectorizer()
tf = vectorizer.fit_transform(docs)
term_freq = pd.DataFrame(tf.toarray(),columns= vectorizer.get_feature_names_out()) 
print(term_freq)
```
