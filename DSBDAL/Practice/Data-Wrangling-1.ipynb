## Data Formatting
Convert data into correct types

```python
# check all column data types
df.dtypes
```

```python
# convert the string dataypes into the category
col = df.select_dtypes(include='str').columns
print(col)
```

```python
# select the Perticular column for conversion
col = ['Gender', 'Married','Education','Self_Employed', 'Property_Area', 'Loan_Status']
```

```python
# conversion
for i in col:
    df[i]=df[i].astype('category')
```

```python
df.dtypes
```

```python
# convert the ApplicantIncome column int to float
df['ApplicantIncome'] = df['ApplicantIncome'].astype('float')
```

```python
df.dtypes
```

```python
# convert the LoanAmount into integer
df['LoanAmount']= df['LoanAmount'].astype('Int64')
```

```python
df.dtypes
```

## Data Normalization
Only numeric columns are normalized


```python
# Show original values
df[['ApplicantIncome', 'LoanAmount']].head()
```

```python
# Using Min-Max Scaling
# formula  x' = (x- x_min)/(x_max - x_min)
df['ApplicantIncome_norm'] = ( df['ApplicantIncome']-df['ApplicantIncome'].min() ) / (df['ApplicantIncome'].max() - df['ApplicantIncome'].min())
```

```python
df[['ApplicantIncome', 'ApplicantIncome_norm','LoanAmount']].head()
```

```python
# Min-Max Scaling Using sklearn
from sklearn.preprocessing import MinMaxScaler
scaler= MinMaxScaler()
df['LoanAmount_norm']= scaler.fit_transform(df[['LoanAmount']])
```

```python
df[['ApplicantIncome', 'ApplicantIncome_norm','LoanAmount','LoanAmount_norm']].head()
```

```python
# Z-score normalization
# formula : x' = x - mean / std

col = 'LoanAmount'

mean = df[col].mean()
std = df[col].std()

df[col + "_zscore"] = (df[col] - mean) / std

df.head()
```

## Convert Categorical Variables

```python
# Label Encoding using map function
df['Gender'] = df['Gender'].map({ 'Male' : 'M', 'Female' :'F'})
```

```python
df.head()
```

```python
# Checking the Unique Values
df['Married'].unique()
```

```python
df['Married_Encodeing']= df['Married'].map({'Yes':'Y','No':'N'})
```

```python
df.head()
```

```python
# Convert categorical columns using one-hot encoding

df = pd.get_dummies(df,columns=['Property_Area'],prefix='Property')
df.head()
```
