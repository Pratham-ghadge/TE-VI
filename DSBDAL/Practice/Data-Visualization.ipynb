```python
# Distribution of passenger class
sns.histplot(df['pclass'], kde=True, bins=3)
plt.xticks([1,2,3])
plt.title("Distribution of Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()
```

```python
# Distribution of embark town
sns.histplot(df['embark_town'] )
plt.title("Distribution of Embark Town")
plt.xlabel("Town")
plt.ylabel("Count")
plt.show()
```

```python
# Subplot -> subplot is use for creating the two or more subplots inside the one frame used for comparison side by side plottingb graph
# Create a figure with a fixed size (width=12, height=5)
plt.figure(figsize=(12, 5))

# ------------------ SUBPLOT 1 ------------------
# Create first subplot (1 row, 2 columns, position 1)
plt.subplot(1, 2, 1)

# Plot histogram for passenger class with KDE curve
sns.histplot(df['pclass'], bins=3, kde=True)

# Set x-axis ticks to show only class values (1, 2, 3)
plt.xticks([1, 2, 3])

# Add title and labels
plt.title("Distribution of Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")

# ------------------ SUBPLOT 2 ------------------
# Create second subplot (1 row, 2 columns, position 2)
plt.subplot(1, 2, 2)

# Plot countplot for embark town (categorical data)
sns.countplot(x='embark_town', data=df)

# Add title and labels
plt.title("Distribution of Embark Town")
plt.xlabel("Town")
plt.ylabel("Count")

# Adjust layout to prevent overlap between plots
plt.tight_layout()

# Display both plots together
plt.show()
```

```python
correlation = df[['survived','pclass','sibsp','parch','fare']]
```

```python
correlation.corr()
```

```python
sns.heatmap(correlation.corr(),annot=True)
```

```python
# Create a cross-tab (matrix)
pclass_survival = pd.crosstab(df['pclass'], df['survived'])

# Plot heatmap
sns.heatmap(pclass_survival, annot=True, fmt='d', cmap='Blues')
plt.title('Heatmap of Passenger Class vs Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Passenger Class')
```
