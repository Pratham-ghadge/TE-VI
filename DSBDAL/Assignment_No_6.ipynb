```python
# Feature List and Types
col = df.columns
dtype = df.dtypes

print("Dataset Features And Types:\n")

for feature, type in zip(col, dtype):
    print(f"{feature} --> {type}")

# Observation
# The Iris dataset contains 4 numeric features (float values).
# The species column is categorical (object type).
# Numeric features represent flower measurements, while species is a nominal label.
```

```python
# Histogram for Each Feature
df.hist(figsize=(10, 8), edgecolor='black')
plt.suptitle("Histogram of Iris Dataset Features", fontsize=16)
plt.show()

# Observations 
# Sepal length is mostly between 5 to 6.5 cm
# Sepal width is concentrated around 3 cm
# Petal length and petal width show clear variation among species
# Petal features are more useful for classification
```

```python
# Box Plot for Each Feature

plt.figure(figsize=(12, 8))

for i, col in enumerate(df.columns[:-1]):
    plt.subplot(2, 2, i+1)
    sns.boxplot(y=df[col])
    plt.title(f"Box Plot of {col}")

plt.tight_layout()
plt.show()

# Observation
# Petal length and petal width have wider spread
# Sepal width contains a few outliers
# Median values differ across features
```

```python
# Box Plot Comparison by Species
plt.figure(figsize=(12, 8))

for i, col in enumerate(df.columns[:-1]):
    plt.subplot(2, 2, i+1)
    sns.boxplot(x="species", y=col, data=df)
    plt.title(f"{col} Distribution by Species")

plt.tight_layout()
plt.show()

# Observation
# Setosa has the smallest petal length and width
# Virginica has the largest petal dimensions
# Petal features clearly separate the species groups
```

```python
# Compare distribution of all features
plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.title("Comparison of Feature Distributions (Iris Dataset)")
plt.show()
# Observation 
# This box plot compares all numeric features together.
# Petal length and petal width have a wider spread.
# Sepal width has some points outside the box → those are outliers.
```

```python
# Checking outliers feature-wise

features = df.columns[:-1]  # numeric columns only

for col in features:
    print("\nFeature:", col)

    Q1 = df[col].quantile(0.25)   # lower range
    Q3 = df[col].quantile(0.75)   # upper range

    IQR = Q3 - Q1                 # range difference

    # Outlier limits
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Outlier values
    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print("Number of Outliers:", outliers.shape[0])
```
