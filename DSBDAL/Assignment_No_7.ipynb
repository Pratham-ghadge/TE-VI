```python
# Create Sample Document
document = "Prathamesh is learning Natural Language Processing. He is studying NLP for his college assignment."

print("Original Document:\n")
print(document)
```

```python
# Tokenization
tokens = word_tokenize(document)

print("Tokens:\n")
print(tokens)

# Observation:
# The sentence is broken into individual words and punctuation marks.
```

```python
# POS Tagging
pos_tags = pos_tag(tokens)

print("POS Tags:\n")
print(pos_tags)

# Observation:
# Each word is assigned a grammatical tag such as noun, verb, adjective etc.
```

```python
# Stopwords Removal
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in tokens:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print("After Stopword Removal:\n")
print(filtered_words)

# Observation:
# Common words like 'is', 'for', 'his' are removed to keep meaningful words.
```

```python
# Stemming
stemmer = PorterStemmer()

stemmed_words = []

for word in filtered_words:
    stemmed_words.append(stemmer.stem(word))

print("After Stemming:\n")
print(stemmed_words)

# Observation:
# Words are reduced to root form but may not be proper dictionary words.
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = []

for word in filtered_words:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print("After Lemmatization:\n")
print(lemmatized_words)

# Observation:
# Words are converted into meaningful base form.
```

```python
# Term Frequency Calculation

tf_dict = {}
total_words = len(lemmatized_words)

for word in lemmatized_words:
    if word in tf_dict:
        tf_dict[word] += 1
    else:
        tf_dict[word] = 1

print("Term Frequency:\n")

for word in tf_dict:
    tf = tf_dict[word] / total_words
    print(word, "=", tf)

# Observation:
# TF shows how frequently a word appears in the document.
```

```python
# Multiple Documents for IDF Calculation
documents = [
    "NLP assignment is interesting",
    "Machine learning and NLP project",
    "College assignment and project"
]

total_documents = len(documents)

print("IDF Values:\n")

for word in tf_dict:
    doc_count = 0
    
    for doc in documents:
        if word in doc:
            doc_count += 1
    
    if doc_count > 0:
        idf = math.log(total_documents / doc_count)
        print(word, "=", idf)
    else:
        print(word, "= Not Found")

# Observation:
# IDF reduces importance of common words appearing in multiple documents.
```
