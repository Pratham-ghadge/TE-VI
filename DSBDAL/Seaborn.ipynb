```python
#import the Seaborn library
import seaborn as sns
import pandas as pd 
import numpy as np 
```

```python
#get the list of the data which come with seaborn
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
# check the diffrent kind of species and their count of penguins
penguins['species'].value_counts()
```

```python
#check the diffrent island 
penguins['island'].value_counts()
```

```python
# sea image of penguins and understand dataset and penguins body parts
```

```python
# plot the data using Scatterplot
sns.scatterplot(data = penguins , x = "flipper_length_mm", y = "body_mass_g")
#seaborn allows you to pass the third variable hue to differentiate
sns.scatterplot(data = penguins , x = "flipper_length_mm", y = "body_mass_g", hue = 'island')
```

```python
# we can also set the background of the graph 
# themes availabel white ,dark,whitegrid,darkgrid,ticks
sns.set_style('darkgrid')
sns.scatterplot(data = penguins , x = "flipper_length_mm", y = "body_mass_g",hue ='sex')
```

```python
sns.scatterplot(data = penguins ,x ='species' , y ='body_mass_g', hue='island' )
```

```python
# Stripplot : In a strip plot, the data points are "stripped" along the axis of the category they belong to
sns.set_style('ticks')
sns.stripplot(data = penguins ,x ='species' , y ='body_mass_g', hue='island')
```

```python
# dodge is use to seperate datapoints based on hue
#sns.stripplot(data = penguins ,x ='species' , y ='body_mass_g', hue='island',dodge=True)
#jitter is used to see the override data points based on the shades of data points
sns.stripplot(data = penguins ,x ='species' , y ='body_mass_g', hue='island',dodge=True,jitter=True)
```

```python
#Swarm Plot
sns.swarmplot(data = penguins,x='species',y='body_mass_g',hue ='island')
```

```python
#histogram
sns.histplot(data=penguins ,x = 'body_mass_g' , hue = 'sex')
```

```python
#regression plot
sns.regplot(data=penguins,x ='body_mass_g', y = 'flipper_length_mm')
```

```python
#pair plot
sns.pairplot(penguins,hue='sex')
```

```python
#rugplot
sns.rugplot(penguins)
```

### Categorical Data Plots

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pa
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
sns.countplot(x =df['sex'] , hue=df['smoker'] )
```

```python
#barplot
sns.barplot(x =df['sex'], y = df['total_bill'])
```

```python
sns.barplot(x =df['sex'], y = df['total_bill'], estimator = np.sum)
```

```python
#boxplot
sns.boxplot(x = 'tip' , y = 'day', data = df)
```

```python
#violinplot
sns.violinplot(x = 'day' , y = 'total_bill', data = df)
```
