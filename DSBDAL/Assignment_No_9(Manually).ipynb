```python
# import libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Load dataset
data = pd.read_csv("Social_Network_Ads.csv")
```

```python
# Select features
X = data[['Age','EstimatedSalary']].values
y = data['Purchased'].values
```

```python
# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0
)
```

```python
# Feature Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# -------------------------------
# Logistic Regression Class
# -------------------------------

class LogisticRegressionManual:

    def __init__(self):
        self.weights = None
        self.bias = None

    # 🔹 Sigmoid Function
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    # 🔹 Forward Propagation
    def forward(self, X):
        z = np.dot(X, self.weights) + self.bias
        return self.sigmoid(z)

    # 🔹 Cost Function (Log Loss)
    def cost(self, y_pred, y):
        n = len(y)
        return -np.mean(y * np.log(y_pred + 1e-9) + (1 - y) * np.log(1 - y_pred + 1e-9))

    # 🔹 Backward Propagation
    def backward(self, X, y, y_pred):
        n = X.shape[0]

        dw = (1/n) * np.dot(X.T, (y_pred - y))
        db = (1/n) * np.sum(y_pred - y)

        return dw, db

    # 🔹 Update Parameters
    def update(self, dw, db, lr):
        self.weights -= lr * dw
        self.bias -= lr * db

    # 🔹 Training
    def train(self, X, y, lr=0.01, iters=1000):
        n_features = X.shape[1]

        self.weights = np.zeros(n_features)
        self.bias = 0

        for i in range(iters):
            y_pred = self.forward(X)

            cost = self.cost(y_pred, y)

            dw, db = self.backward(X, y, y_pred)

            self.update(dw, db, lr)

            if i % 100 == 0:
                print(f"Iteration {i}, Loss = {cost}")

    # 🔹 Prediction
    def predict(self, X):
        probs = self.forward(X)
        return [1 if i >= 0.5 else 0 for i in probs]

```

```python
# Train Model
model = LogisticRegressionManual()
model.train(X_train, y_train, lr=0.01, iters=2000)
```

```python
# Prediction
y_pred = model.predict(X_test)
```

```python
# Confusion Matrix (Manual)

TP = TN = FP = FN = 0
for i in range(len(y_test)):
    if y_test[i] == 1 and y_pred[i] == 1:
        TP += 1
    elif y_test[i] == 0 and y_pred[i] == 0:
        TN += 1
    elif y_test[i] == 0 and y_pred[i] == 1:
        FP += 1
    elif y_test[i] == 1 and y_pred[i] == 0:
        FN += 1

print("\nConfusion Matrix:")
print([[TN, FP], [FN, TP]])
```

```python
# Metrics
accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)

print("\nAccuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
```
