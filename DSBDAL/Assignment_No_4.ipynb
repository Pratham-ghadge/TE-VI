```python
# Importing Libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Show all In-built seaborn dataset
datasets = sns.get_dataset_names()
print("Dataset in Seaborn:")
for data in datasets:
    print(data)
```

```python
# Load Titanic dataset
df= sns.load_dataset('titanic')

# Check first few rows
df.head()
```

```python
# check missing and Null values 
df.isnull().sum()
```

```python
# Count of Male and Female Passengers Traveling on the Titanic
sns.countplot(data=df, x='sex')
plt.title('Number of Male and Female Passengers on Titanic')
plt.xlabel('Gender')
plt.ylabel('Count')
# Observation of below Graph :
# The number of male passengers was higher than female passengers on the Titanic.
```

```python
# Count of Male and Female Passengers Based on Survival Status
sns.countplot(data=df, x='sex', hue='survived')
plt.title('Gender Distribution Based on Survival')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
# Observation
# Female passengers had a higher survival rate compared to males.
```

```python
# Heatmap Showing Survival Count Across Passenger Classes on the Titanic

# Create a cross-tab (matrix)
pclass_survival = pd.crosstab(df['pclass'], df['survived'])

# Plot heatmap
sns.heatmap(pclass_survival, annot=True, fmt='d', cmap='Blues')
plt.title('Heatmap of Passenger Class vs Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Passenger Class')

# Numbers inside the boxes represent passenger counts.
# Darker color → Higher number of passengers.
# Lighter color → Fewer passengers.
```

```python
# Violin Plot Showing Age Distribution of Male and Female Passengers Based on Survival Status
sns.violinplot(data=df, x='sex', y='age', hue='survived', split=True)
plt.title('Age-wise Gender Distribution Based on Survival')
plt.xlabel('Gender')
plt.ylabel('Age')
# Observation:
# passengers have a wider spread in ages.
```

```python
#Swarm Plot Showing Age-wise Distribution of Male and Female Passengers with Survival
sns.swarmplot(data=df, x='sex', y='age', hue='survived')

plt.title('Swarm Plot of Age by Gender and Survival')
plt.xlabel('Gender')
plt.ylabel('Age')
# Observation
# Each dot represents an individual passenger.
# More male dots appear in the non-survived category.
# Female dots are more concentrated in the survived group.
# Younger passengers are more visible among survivors.
# The swarm plot helps identify actual data distribution without aggregation
```

```python
# Categorical Plot Showing Passenger Count by Embarkation Port with Survival Status Across Different Passenger Classes
# Create catplot
sns.catplot(
    data=df,
    x='embarked',
    hue='survived',
    kind='count',
    col='pclass'
)

plt.suptitle('Embarked vs Survival Across Passenger Classes', y=1.05)

# Passengers from higher classes, especially those embarking at Cherbourg, had greater survival rates compared to third-class passengers from Southampton
```

```python
# Pair Plot Showing Relationships Between Numerical Features of the Titanic Dataset Based on Survival
# Select only numeric columns
numeric_df = df[['survived','pclass','age','sibsp','parch','fare']]
# Create pairplot
sns.pairplot(numeric_df, hue='survived')
plt.suptitle('Pair Plot of Numerical Features Based on Survival', y=1.02)
# Observation 
# Pairplot effectively visualizes relationships among numerical variables
```

```python
# Histogram Showing Distribution of Ticket Fare Among Titanic Passengers
sns.histplot(data=df, x='fare', bins=30, kde=True)
plt.title('Distribution of Ticket Fare')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
# Observations:
# Most passengers paid low fares, indicating many travelers were from lower classes.
```

```python

# Box Plot: Age distribution by Gender and Survival
sns.boxplot(data=df, x='sex', y='age', hue='survived')
plt.title("Age Distribution by Gender and Survival Status (Titanic Dataset)")
plt.xlabel("Gender")
plt.ylabel("Age")

# Observations:
# 1. The median age of passengers is similar across both genders.
# 2. Females show a higher survival rate, especially among younger age groups.
# 3. Males have more passengers who did not survive, across almost all age ranges.
# 4. Some outliers are visible, indicating very old passengers were also present.
```
