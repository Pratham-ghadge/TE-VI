```python
dataset["Gender"].mode()[0]
dataset["Gender"].fillna(dataset["Gender"].mode()[0],inplace=True)
```

- **Now we are filling the missing data into all column**

```python
dataset.select_dtypes(include="object").columns
```

```python
dataset.select_dtypes(include="int64").columns
```

```python
dataset.columns
```

```python
for i in dataset.select_dtypes(include="object").columns:
    modes = dataset[i].mode()
    if not modes.empty:
        dataset[i] = dataset[i].fillna(modes[0])
```

```python
dataset.isnull().sum()
```

```python
dataset.select_dtypes(include="float64").columns
```

```python
for i in dataset.select_dtypes(include="float64").columns:
    mean_value=dataset[i].mean()
    dataset[i]= dataset[i].fillna(mean_value)
```

```python
dataset.isnull().sum()
```

```python
dataset.head(10)
```

```python
dataset["Loan_Status"] = dataset["Loan_Status"].map({"Y":1,"N":0})
```

```python
dataset.head(10)
```

**One Hot Encoding**

```python
property_dummies = pd.get_dummies(dataset["Property_Area"],prefix="Property",dtype=int)
```

```python
dataset = pd.concat([dataset,property_dummies],axis=1)
```
