### MATPLOTLIB ###

```python
#import library
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#matplotlib basics
x = np.linspace(0,5,11)
x
y= x**2
y
```

```python
#plot the graph
plt.plot(x ,y)
plt.title("Plot graph")
plt.xlabel("numbers")
plt.ylabel("square")
```

```python
#subplot in matplot
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x**2,x**2)
```

```python
#object Oriented ways
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.4,0.3,0.4])
axis2.plot(x,y)
```

```python
# Types of Plots
plt.scatter(x,y)

```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

```python
#saving plots
fig = plt.figure()
ax =fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
# saving a image 
plt.savefig("basicplot.png")
```
