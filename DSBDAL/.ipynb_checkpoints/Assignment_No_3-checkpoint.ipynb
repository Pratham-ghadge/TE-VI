- **Descriptive Statistic**

  - **Create the group Age Wise**

```python
bins =[0,20,30,40,50,60,100]
labels = [
    "0–20",
    "21–30",
    "31–40",
    "41–50",
    "51–60",
    "61–100"
]



df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=True)

```

```python
std_dev_by_age_range = df.groupby('Age_Group')['Spending Score (1-100)'].std()
print(std_dev_by_age_range )
```

```python
grouped = df.groupby("Age_Group")['Spending Score (1-100)']
grouped.describe()
```

```python
grouped = df.groupby("Age_Group")
grouped.describe().T
```
